# Source / published Dataset exploratory analysis

Inspect the actual published steady/transient neural-operator inputs, states, conditioning and targets, then compare aggregate quality and physical trajectories. Package membership and field contracts remain authoritative.

Generation inputs and candidate outcome accounting belong in [Generation-input EDA](01_eda_generation_inputs.ipynb). Training-fitted normalizers are separate and are not reconstructed here.

`MAX_CASES` limits inspected cases per package. Values precede training-fitted normalization.

## Choose the scientific evidence view

`EDA_VIEW = "published_dataset"` inspects immutable published package membership and model-facing representation. Use `EDA_VIEW = "generated_source"` to inspect actual admitted COMSOL Source outputs before Dataset construction: physical fields, states, schedules and successful/failed/pending case accounting. Source campaigns have no model training, validation, test or OOD memberships. Rerun the workspace cell after changing `EDA_VIEW`.


```python
from IPython.display import display as show

from src import common
from src.analysis import eda

EDA_VIEW = "published_dataset"  # "published_dataset" or "generated_source"
MAX_CASES: int | None = 8
STORAGE_ROOT = common.paths.get_storage_root()
```

```python
DISPLAY_SPATIAL_STRIDE = "auto"  # 1, 2, 4, or auto; maps only, full-grid statistics

if EDA_VIEW == "published_dataset":
    workspace = eda.workspace.prepare_published_dataset_eda_workspace(
        storage_root=STORAGE_ROOT,
        display_spatial_stride=DISPLAY_SPATIAL_STRIDE,
        max_cases=MAX_CASES,
        title="Published Dataset EDA",
    )
elif EDA_VIEW == "generated_source":
    workspace = eda.workspace.prepare_generated_output_eda_workspace(
        storage_root=STORAGE_ROOT,
        display_spatial_stride=DISPLAY_SPATIAL_STRIDE,
        max_cases=MAX_CASES,
        title="Generated Source EDA",
    )
else:
    message = 'EDA_VIEW must be "published_dataset" or "generated_source".'
    raise ValueError(message)
print(workspace.summary_text)
```

```python
if workspace.panel is not None:
    show(workspace.panel)
```
